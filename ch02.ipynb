```python
import re

class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i: s for s, i in vocab.items()}
    
    def encode(self, text):
        preprocessed = re.split(r'([,;:.?!()"\\\'-]|\s|--)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
    
    def decode(self, ids):
        
        text = " ".join([self.int_to_str[i] for i in ids])
        text = re.sub(r'\s+([,;:.?!()"\\\'])', r'\1', text)
        return text
```

# 总结

1、去重：使用 set(preprocessed) 剔除重复 Token

2、排序：使用 sorted() 函数对去重后的 Token 按字母顺序排列

3、用 enumerate() 函数遍历排序后的 Token 列表，生成映射关系，为每个 Token 分配递增的唯一整数 ID，形成 Token → ID 的正向映射

4、对字典进行反向，形成 ID → Token的反向映射

5、简单分词器（SimpleTokenizer）：正向映射：字符串 Token → 整数 ID，反向映射：整数 ID → 字符串 Token

6、encode 方法：基于正向映射，将 Token 列表转换为 Token ID 序列

7、decode 方法：基于反向映射，将 Token ID 序列还原为文本字符串

```python
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i: s for s, i in vocab.items()}
    
    def encode(self, text):
        preprocessed = re.split(r'([,;:.?!()"\\\'-]|\s|--)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        preprocessed = [
            item if item in self.str_to_int 
            else "<|unk|>" for item in preprocessed
            ]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
    
    def decode(self, ids):
        
        text = " ".join([self.int_to_str[i] for i in ids])
        text = re.sub(r'\s+([,;:.?!()"\\\'])', r'\1', text)
        return text
```

# 2.5 字节对编码

- GPT-2 使用字节对编码（BPE）作为其分词器。
- 它允许模型将不在预定义词汇表中的单词拆分为更小的子词单元，甚至是单个字符，从而能够处理未登录词（out-of-vocabulary words）。
- 例如，如果 GPT-2 的词汇表中没有 “unfamiliarword” 这个词，它可能会将其分词为 `["unfam", "iliar", "word"]` 或其他子词拆分形式，具体取决于它训练得到的 BPE 合并规则。
- 原始的 BPE 分词器可以在这里找到：https://github.com/openai/gpt-2/blob/master/src/encoder.py
- 在本章中，我们使用 OpenAI 开源的 `tiktoken` 库中的 BPE 分词器，该库的核心算法用 Rust 实现，以提高计算性能。

```python
# 导入 tiktoken 库，这是 OpenAI 开源的高性能分词器
import tiktoken

# 导入 importlib 模块，用于获取已安装库的版本信息
import importlib

# 打印当前安装的 tiktoken 版本号
print("tiktoken version:", importlib.metadata.version("tiktoken"))
```

```python
# 获取 GPT-2 对应的 BPE 分词器实例
# 'gpt2' 表示使用与 GPT-2 完全相同的分词规则和词汇表
tokenizer = tiktoken.get_encoding('gpt2')
```

```python
# 定义要处理的文本
# <|endoftext|> 是 GPT-2 使用的特殊结束标记，用来分隔不同的文本片段
text = (
    "Hello, do you like tea? <|endoftext|> In the sunlit terraces"
    "of someunknownPlace."
)

integers = tokenizer.encode(text, allowed_special={'<|endoftext|>'})

integers
```

```python
# 将整数 Token ID 解码回原始文本
strings = tokenizer.decode(integers)

strings
```

![](images/bpe分词器原理.png)

BPE 的原理可以概括为：

1. **基础底裤**：自带 256 个单个字符的基础 Token（覆盖所有字节）；
2. **学高频对**：训练时合并文本里的高频字符对，生成子词 Token；
3. **拆未知词**：遇到没见过的词，就用已学的子词 + 基础字符‘拼着拆’，保证能处理任意文本。

1. **未登录词处理**：BPE 通过 “拆为基础字节 + 反向合并已知子词” 识别未知词，核心是 “最小单元兜底，高频组合优化”。
2. **BPE 核心逻辑**：训练时合并高频字节对构建词汇表，编码时按规则合并，解码时反向拆分。
3. **实用选择**：教学用自定义 BPE 实现理解原理，实际项目用`tiktoken`保证性能。

# 2.6 使用滑动窗口进行数据采样

```python
# 以只读模式打开文本文件，编码为UTF-8
with open("the-verdict.txt", "r", encoding="utf-8") as f:
    # 读取文件中的全部原始文本
    raw_text = f.read()
```

```python
# 使用分词器将原始文本编码为模型可处理的数字ID序列
enc_text = tokenizer.encode(raw_text)
# 打印编码后的序列长度，即文本被切分成的token数量
print(len(enc_text))
```

```python
# 从编码后的序列中截取从第50个token开始的片段作为示例
enc_sample = enc_text[50:]
```

```python
# 设置上下文窗口大小为4
context_size = 4

# 取enc_sample的前context_size个token作为输入x
x = enc_sample[:context_size]
# 取enc_sample的第1到context_size+1个token作为目标y（输入向右偏移1位）
y = enc_sample[1:context_size+1]

# 打印输入x和目标y的token序列
print(f"x: {x}")
print(f"y: {y}")
```

```python
# 遍历1到context_size（包含）的范围，展示逐步预测过程
for i in range(1, context_size+1):
    # 取前i个token作为当前上下文
    context = enc_sample[:i]
    # 第i个token是当前上下文需要预测的目标词
    desired = enc_sample[i]

    # 打印上下文和对应的目标token
    print(context, "----->", desired)
```

```python
# 再次遍历，这次把token解码为可读文本
for i in range(1, context_size+1):
    context = enc_sample[:i]
    desired = enc_sample[i]

    # 打印解码后的上下文文本和目标词
    print(tokenizer.decode(context), "----->", tokenizer.decode([desired]))
```

```python
# 导入PyTorch深度学习框架
import torch

# 打印当前PyTorch的版本号，用于确认安装是否成功
print("PyTorch version:", torch.__version__)
```

**我们采用滑动窗口的方式，每次将窗口向右移动 1 个位置：**

![](images/2.13.png)

**创建数据集和数据加载器，从输入文本数据集中提取文本块。**

```python
# 从PyTorch的工具包中导入Dataset和DataLoader，用于构建自定义数据集和数据加载器
from torch.utils.data import Dataset, DataLoader

# 定义一个继承自PyTorch Dataset的自定义数据集类，用于处理GPT模型的训练数据
class GPTDatasetV1(Dataset):
    def __init__(self, txt, tokenizer, max_length, stride):
        # 初始化存储输入和目标token序列的列表
        self.input_ids = []
        self.target_ids = []

        # 将整个输入文本编码为模型可处理的token ID序列
        # allowed_special参数允许包含特殊的结束符"<|endoftext|>"
        token_ids = tokenizer.encode(txt, allowed_special={"<|endoftext|>"})

        # 断言检查：编码后的token数量必须至少比max_length多1
        # 否则无法生成有效的输入-目标对
        assert len(token_ids) > max_length, "编码后的输入数量必须至少等于max_length+1"

        # 使用滑动窗口技术，将长文本切分为多个重叠的max_length长度的序列
        for i in range(0, len(token_ids) - max_length, stride):
            # 截取从i到i+max_length的token作为输入块
            input_chunk = token_ids[i:i + max_length]
            # 截取从i+1到i+max_length+1的token作为目标块（输入向右偏移1位）
            target_chunk = token_ids[i + 1: i + max_length + 1]

            # 将输入块和目标块转换为PyTorch张量，并添加到对应的列表中
            self.input_ids.append(torch.tensor(input_chunk))
            self.target_ids.append(torch.tensor(target_chunk))

    # 返回数据集中样本的总数
    def __len__(self):
        return len(self.input_ids)

    # 根据索引idx获取对应的输入和目标张量对
    def __getitem__(self, idx):
        return self.input_ids[idx], self.target_ids[idx]
```

```python
def create_dataloader_v1(txt, batch_size=4, max_length=256,
                         stride=128, shuffle=True, drop_last=True,
                         num_workers=0):
    # 初始化分词器：使用tiktoken加载GPT-2的分词器
    tokenizer = tiktoken.get_encoding("gpt2")

    # 创建数据集：使用我们之前定义的GPTDatasetV1类
    dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)

    # 创建数据加载器：将数据集包装成可迭代的批量加载器
    dataloader = DataLoader(
        dataset,
        batch_size=batch_size,      # 每个批次包含的样本数
        shuffle=shuffle,            # 是否在每个epoch打乱数据顺序
        drop_last=drop_last,        # 是否丢弃最后一个不完整的批次
        num_workers=num_workers     # 用于加载数据的子进程数量
    )

    return dataloader
```

```python
# 打开文本文件并读取内容
with open("the-verdict.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()
```

```python
# 创建数据加载器
# 传入原始文本，并设置批量大小为1，上下文窗口长度为4，步幅为1，不打乱数据顺序
dataloader = create_dataloader_v1(
    raw_text, batch_size=1, max_length=4, stride=1, shuffle=False
)

# 将数据加载器转为迭代器，以便逐个获取批次
data_iter = iter(dataloader)
# 获取第一个批次
first_batch = next(data_iter)
# 打印第一个批次的输入和目标张量
print(first_batch)
```

```python
# 获取第二个批次
second_batch = next(data_iter)
# 打印第二个批次的输入和目标张量
print(second_batch)
```

```python
# 创建数据加载器
# 设置批量大小为8，上下文窗口长度为4，步幅为4（与窗口长度相同，无重叠），不打乱数据顺序
dataloader = create_dataloader_v1(raw_text, batch_size=8, max_length=4, stride=4, shuffle=False)

# 将数据加载器转为迭代器
data_iter = iter(dataloader)
# 获取第一个批次的输入和目标张量
inputs, targets = next(data_iter)

# 打印输入张量
print("Inputs:\n", inputs)
# 打印目标张量
print("\nTargets:\n", targets)
```

关键参数 ** 步幅（stride）** 决定样本重叠程度：步幅等于窗口大小时样本无重叠，可降低过拟合；步幅小于窗口大小时样本重叠，适用于小数据集扩充样本量。

# 2.7 创建词嵌入

![](images/2.15.png)

- 假设我们有四个输入示例，经过分词后得到的输入 ID 是 2、3、5 和 1：
- 为了简化说明，我们假设词汇表很小，只有 6 个词，并且希望创建维度为 3 的嵌入向量：
- 这会生成一个 6×3 的权重矩阵：

```python
import torch

# 1. 定义输入的token ID（分词后的结果）
# 输入序列包含四个token，对应的ID分别是2、3、5、1
input_ids = torch.tensor([2, 3, 5, 1])

# 2. 定义嵌入层的参数
vocab_size = 6    # 词汇表大小，总共有6个不同的词
output_dim = 3    # 每个词的嵌入向量维度是3

# 3. 设置随机种子，保证每次运行结果一致
torch.manual_seed(123)

# 4. 创建嵌入层
# Embedding层会自动初始化一个 vocab_size × output_dim 的权重矩阵
embedding_layer = torch.nn.Embedding(vocab_size, output_dim)

# 5. 打印嵌入层的权重矩阵
# 这个矩阵里的每一行对应一个词的嵌入向量，会在训练中不断更新
print(embedding_layer.weight)
```

- 嵌入层本质上就是一个**查表（look-up）操作**

```python
# 1. 对单个token ID（3）进行嵌入
# 输入是一个包含ID=3的张量，输出就是权重矩阵中第3行对应的3维向量
print(embedding_layer(torch.tensor([3])))
```

```python
# 2. 对整个输入序列的token IDs进行嵌入
# 输入是[2, 3, 5, 1]，输出是每个ID对应的嵌入向量组成的矩阵
print(embedding_layer(input_ids))
```

**嵌入层本质上就是一个查表操作**：

输入文本 `fox jumps over dog` 先被分词并映射为 token ID `[2, 3, 5, 1]`。

嵌入层的权重矩阵是一个 6×3 的表，每个 token ID 对应矩阵里的一行。

当我们需要嵌入时，只需根据 ID 去矩阵中 “查” 对应的行，就能得到该 token 的连续向量表示。

## 嵌入层（Embedding Layer） vs 线性层（Linear Layer）

|     特性     |                  嵌入层（Embedding Layer）                   |                    线性层（Linear Layer）                    |
| :----------: | :----------------------------------------------------------: | :----------------------------------------------------------: |
|   **目标**   | 把离散的类别 / 索引（如词的 ID）映射到低维连续向量空间，捕捉语义关联 | 对连续向量做线性变换（`y = xW + b`），用于特征提取或维度转换 |
| **输入类型** |       离散的整数索引（如 `input_ids = [2, 3, 5, 1]`）        |         连续的浮点向量（如 `x = [0.1, 0.5, -0.3]`）          |
| **典型场景** |             NLP 中的词嵌入、推荐系统中的物品嵌入             | 分类 / 回归任务的输出层、Transformer 中的前馈网络、特征维度变换 |

- **嵌入层**
    - 本质是一个**查找表（lookup table）**，权重矩阵的形状是 `[vocab_size, output_dim]`。
    - 输入一个索引 `i`，就直接取出矩阵的第 `i` 行作为输出，不涉及任何计算。
    - 例如：输入 `input_ids = [2, 3]`，就会取出权重矩阵的第 2 行和第 3 行作为嵌入向量。
- **线性层**
    - 本质是一个**矩阵乘法 + 偏置**运算，权重矩阵的形状是 `[input_dim, output_dim]`。
    - 输入一个向量 `x`，会计算 `y = x @ weight + bias`，是对输入的全局变换。
    - 例如：输入 `x = [1.0, 2.0, 3.0]`，会与权重矩阵做乘法得到输出向量。

# 总结

词元嵌入的作用是将**离散的 token ID**（如分词后的`[2,3,5,1]`）映射为**低维连续向量**，用于大语言模型的输入表示。

嵌入层的本质是**查表操作**：权重矩阵形状为`[vocab_size, output_dim]`，输入 token ID 对应矩阵的行索引，直接取出该行作为嵌入向量。

输入必须是**整数类型的张量**（token ID 序列），输出为对应维度的连续向量矩阵。

嵌入层权重默认可训练（`requires_grad=True`），支持反向传播更新。

# 2.8 编码词的位置

嵌入层会将词 ID 转换为完全相同的向量表示，无论该词在输入序列中的位置如何。

在示例句子 fox jumps over fox 中：

两个 fox 对应的 ID 都是 2

经过嵌入层后，它们得到的向量完全相同（都是 [1.2753, -0.2010, -0.1606]）

这说明普通的嵌入层不包含位置信息，模型无法区分同一个词在句子中的不同位置

位置嵌入会与词嵌入向量组合，形成大语言模型的输入嵌入。

Token embeddings（词嵌入）：只包含词本身的语义信息，同一个词的向量完全相同。

Positional embeddings（位置嵌入）：只包含词在序列中的位置信息，比如第 1 个位置、第 2 个位置的向量各不相同。

Input embeddings（输入嵌入）：将词嵌入和位置嵌入相加，得到的向量同时包含了语义信息和位置信息。

第 1 个 fox 的输入嵌入 = 词嵌入[fox] + 位置嵌入[第1位]

第 4 个 fox 的输入嵌入 = 词嵌入[fox] + 位置嵌入[第4位]

这样模型就能区分这两个 fox 在句子中的不同位置了。

```python
import torch
import torch.nn as nn

# 1. 定义超参数
vocab_size = 10000  # 词汇表大小
embed_dim = 512     # 嵌入向量的维度
max_seq_len = 512   # 最大序列长度
```

```python
# 2. 定义嵌入层
token_embedding = nn.Embedding(vocab_size, embed_dim)  # 词嵌入层
pos_embedding = nn.Embedding(max_seq_len, embed_dim)   # 位置嵌入层
```

```python
# 3. 构造输入
# 假设输入序列是 "fox jumps over fox"，对应的ID为 [2, 3, 5, 2]
input_ids = torch.tensor([2, 3, 5, 2])  # 词ID
seq_len = input_ids.size(0)             # 序列长度（这里是4）
positions = torch.arange(seq_len)       # 位置索引 [0, 1, 2, 3]
```

```python
# 4. 计算嵌入
token_embeds = token_embedding(input_ids)  # 词嵌入，shape: [4, 512]
pos_embeds = pos_embedding(positions)      # 位置嵌入，shape: [4, 512]
```

```python
# 5. 组合得到输入嵌入
input_embeds = token_embeds + pos_embeds   # 相加，shape: [4, 512]
```

```python
# 6. 验证效果
print("第一个fox的词嵌入：", token_embeds[0][:5])  # 取前5个维度查看
print("第四个fox的词嵌入：", token_embeds[3][:5])
print("第一个fox的位置嵌入：", pos_embeds[0][:5])
print("第四个fox的位置嵌入：", pos_embeds[3][:5])
print("第一个fox的输入嵌入：", input_embeds[0][:5])
print("第四个fox的输入嵌入：", input_embeds[3][:5])
```

GPT-2 使用绝对位置嵌入，因此我们只需要再创建一个嵌入层即可。

绝对位置嵌入是给每个位置分配一个固定的向量，模型通过这个向量来区分不同位置的 token。

```python
context_length = max_length # 上下文长度等于最大序列长度（这里是 4）
pos_embedding_layer = torch.nn.Embedding(context_length, output_dim)
# 创建位置嵌入层，输入是位置索引（0 到 3），输出是 256 维向量
```

# 总结

1. 嵌入层基础：词嵌入层将 token ID 转为固定维度向量（如 256 维），仅包含语义信息，同 token 在不同位置的向量完全相同；GPT-2 采用绝对位置嵌入，通过独立的位置嵌入层为每个位置分配固定向量。
2. 输入嵌入生成：将词嵌入与位置嵌入相加，得到同时包含语义和位置信息的输入嵌入，最终形状为[批次大小, 序列长度, 嵌入维度]（如[8, 4, 256]）。
3. GPT 输入流程：原始文本→切分 token→转 token ID→生成词 / 位置嵌入→相加得到输入嵌入→送入仅解码器的 Transformer 模型→输出转换为可读文本。
